```python
# Load the dataset

df = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")

# Display first 5 rows
df.head()
```

```python
# Display the number of rows and columns

print("Shape of dataset:", df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Display first 5 rows
print("\nFirst 5 rows:")
display(df.head())
```

```python
# Display the number of rows and columns

print("Shape of dataset:", df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Display first 5 rows
print("\nFirst 5 rows:")
display(df.head())
```

```python
# Check data types

print(df.dtypes)
```

```python
# Remove leading/trailing spaces from column names

df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
# Check missing values in each column

print(df.isnull().sum())
```

```python
# Missing-value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
# Convert Date column to datetime format

df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# Check the result

print(df["Date"].dtype)
print(df["Date"].head())
```

```python
# Columns that should contain numerical data

numeric_columns = [
    "Estimated Unemployment Rate (%)",
    "Estimated Employed",
    "Estimated Labour Participation Rate (%)"
]

# Convert them to numeric values
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Check data types again
print(df[numeric_columns].dtypes)
```

```python
# Remove columns that are not required for our analysis

columns_to_drop = ["longitude", "latitude"]

df = df.drop(columns=columns_to_drop, errors="ignore")

# Display cleaned dataset
df.head()
```

```python
# Check missing values again

print(df.isnull().sum())
```

```python
# Remove rows containing missing values

df = df.dropna().reset_index(drop=True)

print("Shape after cleaning:", df.shape)
```

```python
# Final inspection

print("Dataset shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

display(df.head())
```

```python
# Create year and month columns

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.strftime("%B")

# Display the new columns

df[["Date", "Year", "Month", "Month_Name"]].head()
```

```python
# Calculate average unemployment rate for each region

region_avg = (
    df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

# Display results

display(region_avg)
```

```python
# Calculate average unemployment rate by month

monthly_avg = (
    df.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_index()
)

display(monthly_avg)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_avg.index,
    monthly_avg.values,
    marker="o"
)

plt.title("Monthly Average Unemployment Rate in India")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Check available regions

print(df["Region"].unique())
```

```python
major_states = [
    "Karnataka",
    "Tamil Nadu",
    "Maharashtra"
]

state_data = df[df["Region"].isin(major_states)]

print(state_data["Region"].unique())
```

```python
plt.figure(figsize=(14, 7))

for state in major_states:
    state_subset = state_data[state_data["Region"] == state]

    plt.plot(
        state_subset["Date"],
        state_subset["Estimated Unemployment Rate (%)"],
        marker="o",
        label=state
    )

plt.title("Unemployment Rate Over Time – Selected Major States")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.legend()
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Calculate average unemployment rate by region

top_10 = (
    df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

display(top_10)
```

```python
plt.figure(figsize=(12, 7))

top_10.sort_values().plot(kind="barh")

plt.title("Top 10 Regions with Highest Average Unemployment Rate")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.tight_layout()
plt.show()
```

```python
# Select columns for correlation analysis

correlation_columns = [
    "Estimated Unemployment Rate (%)",
    "Estimated Employed",
    "Estimated Labour Participation Rate (%)"
]

correlation_matrix = df[correlation_columns].corr()

display(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Unemployment, Employment and Labour Participation")

plt.tight_layout()
plt.show()
```

```python
# Define the COVID-19 split date

covid_date = pd.Timestamp("2020-03-01")

# Create a period column

df["COVID_Period"] = np.where(
    df["Date"] < covid_date,
    "Pre-COVID",
    "Post-COVID"
)

# Check number of observations in each period

print(df["COVID_Period"].value_counts())
```

```python
# Calculate average values for each period

covid_comparison = (
    df.groupby("COVID_Period")[
        [
            "Estimated Unemployment Rate (%)",
            "Estimated Employed",
            "Estimated Labour Participation Rate (%)"
        ]
    ]
    .mean()
)

display(covid_comparison)
```

```python
covid_comparison.plot(
    kind="bar",
    figsize=(12, 7)
)

plt.title("Pre-COVID vs Post-COVID Comparison")
plt.xlabel("Period")
plt.ylabel("Average Value")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

unemployment_comparison = (
    df.groupby("COVID_Period")["Estimated Unemployment Rate (%)"]
    .mean()
)

unemployment_comparison.plot(kind="bar")

plt.title("Average Unemployment Rate: Pre-COVID vs Post-COVID")
plt.xlabel("Period")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
pre_covid = unemployment_comparison["Pre-COVID"]
post_covid = unemployment_comparison["Post-COVID"]

percentage_change = ((post_covid - pre_covid) / pre_covid) * 100

print(f"Pre-COVID average unemployment rate: {pre_covid:.2f}%")
print(f"Post-COVID average unemployment rate: {post_covid:.2f}%")
print(f"Percentage change: {percentage_change:.2f}%")
```
